```python
#ข้อ ป.1
scores = [78, 92, 65, 88, 45, 95, 72]

if len(scores) > 0:
    print("จำนวนผู้เข้าสอบ :", len(scores))
    print("คะแนนสูงสุด :", max(scores))
    print("คะแนนต่ำสูง :", min(scores))
    print("ค่าเฉลี่ย :", round(sum(scores) / len(scores), 2))

    over80 = 0
    failed = []
    for s in scores :
        if s >= 80:
            over80 = over80 + 1
        if s < 50:
            failed.append(s)
    print("ได้ 80 ขึ้นไป :", over80, "คน")
    print("ไม่ผ่านเกณฑ์ 50 :", failed)
else:
    print("ไม่มีข้อมูลคะแนน")
```

```python
#ข้อ ป.2
ids = ["C001", "C002",  "C001", "C003", "C002", "C004"]

total = len(ids)
unique_count = len(set(ids))
dup_count = total - unique_count

print("จำนวนรายการทั้งหมด :", total)
print("จำนวนรหัสที่ไม่ซ้ำ :", unique_count)
print("จำนวนรายการที่ซ้ำ :", dup_count)

unique_ordered = []
for i in ids:
      if i not in unique_ordered:
          unique_ordered.append(i)
print("list ไม่ซ้ำ (คงลำดับ) :", unique_ordered)
```

```python
#ข้อ ป.3
payments = ["cash", "credit", "cash", "transfer", "cash", "credit"]

counter = {}
for p in payments:
    counter[p] = counter.get(p, 0) + 1

for method in sorted(counter):
    print(method, ":", counter[method], "ครั้ง")

best_method = ""
best_count = 0
for method, n in counter.items():
    if n > best_count:
        best_count = n
        best_method = method
print("ใช้มากที่สุด :", best_method, "(", best_count, "ครั้ง )")
```

```python
#ข้อ ป.4
orders = [
{"id": 1001, "customer": "Alice", "city": "Bangkok", "product": "Notebook", "qty": 2, "price":120},
{"id": 1002, "customer": "Bob", "city": "Bangkok", "product": "Pen", "qty": 5, "price": 15},
{"id": 1003, "customer": "Charlie", "city": "Chiang Mai", "product": "Backpack", "qty": 1, "price":850},
{"id": 1004, "customer": "Diana", "city": "Phuket", "product": "Notebook", "qty": 3, "price":120},
{"id": 1005, "customer": "Eve", "city": "Bangkok", "product": "Pen", "qty": 10, "price": 15},
{"id": 1006, "customer": "Frank", "city": "Phuket", "product": "Backpack", "qty": 1, "price": 850},
]

grand_total = 0
for o in orders:
    grand_total = grand_total + o["qty"] * o["price"]
print("1) ยอดขายรวมทั้งหมด :", format(grand_total, ","), "บาท")

sales_by_product = {}
for o in orders:
    name = o["product"]
    amount = o["qty"] * o["price"]
    sales_by_product[name] = sales_by_product.get(name, 0) + amount
print("2) ยอดขายแยกตามสินค้า :")
for name in sorted(sales_by_product):
    print("  ", name, "-", format(sales_by_product[name], ","), "บาท")

customers = []
for o in orders:
    customers.append(o["customer"])
print("3) จำนวนลูกค้าที่ไม่ซ้ำ :", len(set(customers)), "คน")

pairs = []
for o in orders:
    pairs.append((o["qty"] * o["price"], o["customer"]))
pairs.sort(reverse=True)
print("4) สามอันดับยอดสูงสุด :")
for amount, name in pairs[:3]:
    print(" ", name, "-", format(amount, ","), "บาท")
```

```python
#ข้อ ป.5
attendance = {
    "Alice": [1, 2, 3],
    "Bob": [1, 3],
    "Charlie": [2],
    "Diana": [],
    "Eve": [],
}

print("Alice เข้าเรียน", len(attendance.get("Alice", [])), "ครั้ง")
never = []
for name, weeks in attendance.items():
    if len(weeks) == 0:
        never.append(name)
print("คนที่ไม่เคยเข้าเรียน :", never)

#ตามตารางตัดสินใจในส่วนที่ 9.2 ข้อมูลแต่ละช่องมี "ชื่อ" ของตัวเอง (ชื่อนักศึกษา) จึงเลือก dict ก่อน
#ค่าเป็น list เพราะต้องเก็บสัปดาห์ที่เข้าเรียนหลายค่า และมีลำดับ นับด้วย len() ได้ทันที
#นักศึกษาที่ไม่เคยมาเรียนมีคีย์อยู่ แต่ list ว่าง จึงตรวจเจอได้ง่าย ถ้าเก็บเฉพาะคนที่มาเรียน จะหาไม่เจอว่าใครหายไป
#ไม่ใช้ set เพราะ set ไม่รู้จำนวนครั้ง และไม่ใช้หลาย list แยกกัน เพราะเสี่ยงที่ข้อมูลจะไม่ตรงกัน
```
