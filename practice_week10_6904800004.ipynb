```python
orders = [
    {"id": 2001, "customer": "Anan",  "city": "Korat", "product": "Keyboard", "qty": 2, "price": 450},
    {"id": 2002, "customer": "Benja", "city": "Bangkok", "product": "Mouse", "qty": 5, "price": 180},
    {"id": 2003, "customer": "Chai", "city": "Bangkok", "product": "Monitor", "qty": 1, "price": 3200},
    {"id": 2004, "customer": "Duang", "city": "Phuket", "product": "Mouse", "qty": 3, "price": 180},
    {"id": 2005, "customer": "Anan", "city": "Korat", "product": "Monitor",  "qty": 2, "price": 3200},
    {"id": 2006, "customer": "Ekk", "city": "Bangkok", "product": "Keyboard", "qty": 4, "price": 450},
]

VAT_RATE = 0.07

sales = []

for o in orders:
    total = o["qty"] * o["price"]
    vat = total * VAT_RATE
    net = total + vat
    sales.append((o["id"], total, net))

print("=" * 50)
print("รายงานยอดขาย")
print("=" * 50)
print(f"{'รหัสสั่งซื้อ':<12}{'ราคารวม':>15}{'รวม VAT 7%':>18}")
print("-" * 50)

sum_total = 0
sum_net = 0

for order_id, total, net in sales:
  print(f"{order_id:<12}{total:>15,.2f}{net:>18,.2f}")
  sum_total += total
  sum_net += net

print("-" * 50)
print(f"{'ยอดขายรวม':<12}{sum_total:>15,.2f}{sum_net:>18,.2f}")
print("-" * 50)
```
